
## Lab 1 - part 1
### Building The First Agent -  No Framework - Super Simple!

A simple agent who can respond to a simple question.<br>
In this section we're using one of the cheapest LLMs since this is not a complicated task.<br>
The cheaper models have a simpler architecture and they use less neural-network parameters/cells, layers, neurons, or FLOPs.

Documentation:
https://developers.openai.com/api/reference/resources/chat/subresources/completions/methods/create<br>
https://developers.openai.com/api/docs/guides/conversation-state?api-mode=responses


```python
# Create a list of messages in the familiar OpenAI format
# This role is the user; which is the person who is asking the question

messages = [{"role": "user", "content": "Tell me a fun fact"}]
```

```python
messages
```

```python
# And now call it!
# This uses GPT 5.4 nano, an incredibly cheap model
# Here is where you can see the pricings: https://developers.openai.com/api/docs/pricing


response = openai.chat.completions.create(model="gpt-5.4-nano", messages=messages)
print(response.choices[0].message.content)
```

**Also,** we are switching to the next-level LLM because not only will we be asking a question that requires a more detailed explanation, but the response will also be more complex than the previous one.

```python
# And now - let's ask for a question:

question = "Please propose a challenging question to assess someone's Wi-Fi knowledge. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
# ask it - this uses GPT 5.4 mini, still cheap but more powerful than nano

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
# form a new messages list

messages = [{"role": "user", "content": question}]
messages
```

In this section we're feeding the question the second agent created back to itself to get the response.

```python
# Ask the model to answer the hard question

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
# To display the answer more clearly

from IPython.display import Markdown, display

display(Markdown(answer))
```


## Lab 1 - part 3
### More work around the agent; continuing!

This section feeds, is the conclusion of the previous ones.<br>
We're feeding the question and the answer.<br>
Then we check if it was correct.

```python
message = f"""
Here is a question:
{question}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""

print(message)
```

```python
messages = [{"role": "user", "content": message}]
response = openai.chat.completions.create(model="gpt-5.4", messages=messages)
print(response.choices[0].message.content)
```
